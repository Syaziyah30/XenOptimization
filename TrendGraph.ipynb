# C10A1_1_C101C_MTCHW_CH01_CL1.PV

```python
df = pd.read_excel(r"D:\XenOptimization\Data\graph_targetvar.xlsx")
```

```python
df.columns
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Variable to plot
variable = "C10A1_1_C101C_MTCHW_CH01_CL1.PV"

# Get the 5 unique sources
sources = sorted(df["SOURCE"].dropna().unique())

print("Sources:", sources)

# Generate one graph for each source
for source in sources:

    # Filter data for the current source
    dftrend = df.loc[
        df["SOURCE"] == source,
        ["DateTime", variable]
    ].copy()

    # Convert values to numeric
    dftrend[variable] = pd.to_numeric(
        dftrend[variable], errors="coerce"
    )

    # Remove missing values and sort by DateTime
    dftrend = dftrend.dropna(
        subset=["DateTime", variable]
    ).sort_values("DateTime")

    # Create the graph
    plt.figure(figsize=(14, 5))

    # Plot the actual trend
    plt.plot(
        dftrend["DateTime"],
        dftrend[variable],
        linewidth=0.8,
        label="Actual Value"
    )

    # Add red horizontal reference line at 7741.63
    plt.axhline(
        y=7741.63,
        color="red",
        linestyle="-",
        linewidth=1.5,
        label="Reference Value = 7741.63"
    )

    # Graph formatting
    plt.title(f"Trend Graph - {variable} - {source}")
    plt.xlabel("DateTime")
    plt.ylabel(variable)
    plt.xticks(rotation=45)
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

# Combined Trend Graph - 5 sources per variable

```python
import pandas as pd
import matplotlib.pyplot as plt

# Variable and its reference value (red line)
reference_values = {
    "C10A1_1_C101C_MTCHW_CH01_CL1.PV": 7741.63,
    "C10A1_1_C101C_MTCHW_CH01_EF.PV": 85.06,
    "C10A1_1_C101C_MTCHW_CH01_EF.PV2": 0.4107,
    "C10A1_1_C101C_MTCHW_CH01_FIT1.PV": 1343.68,
    "C10A1_1_C101C_MTCHW_CH01_FIT2.PV": 1178.06,
    "C10A1_1_C101C_MTCHW_CH01_TIT1.PV": 30.43,
    "C10A1_1_C101C_MTCHW_CH01_TIT2.PV": 35.89,
    "C10A1_1_C101C_MTCHW_CH01_TIT3.PV": 12.81,
    "C10A1_1_C101C_MTCHW_CH01_TIT4.PV": 18.47,
}

# Get the 5 unique sources
sources = sorted(df["SOURCE"].dropna().unique())

# One figure per variable, with the 5 sources side by side
for variable, ref_value in reference_values.items():

    fig, axes = plt.subplots(
        1, len(sources),
        figsize=(22, 5),
        sharey=True
    )

    for ax, source in zip(axes, sources):

        # Filter data for the current source
        dftrend = df.loc[
            df["SOURCE"] == source,
            ["DateTime", variable]
        ].copy()

        # Convert values to numeric
        dftrend["DateTime"] = pd.to_datetime(dftrend["DateTime"])
        dftrend[variable] = pd.to_numeric(
            dftrend[variable], errors="coerce"
        )

        # Remove missing values and sort by DateTime
        dftrend = dftrend.dropna(
            subset=["DateTime", variable]
        ).sort_values("DateTime")

        # Plot the actual trend
        ax.plot(
            dftrend["DateTime"],
            dftrend[variable],
            linewidth=0.8,
            label="Actual Value"
        )

        # Add red horizontal reference line
        ax.axhline(
            y=ref_value,
            color="red",
            linestyle="-",
            linewidth=1.5,
            label=f"Reference Value = {ref_value}"
        )

        # Subplot formatting
        ax.set_title(f"{source}\n{dftrend['DateTime'].min():%d %b %Y} - {dftrend['DateTime'].max():%d %b %Y}")
        ax.tick_params(axis="x", rotation=45, labelsize=8)
        ax.grid(True, alpha=0.3)

    # Graph formatting
    axes[0].set_ylabel(variable)
    axes[0].legend(loc="best", fontsize=8)
    fig.suptitle(f"Trend Graph - {variable}", fontsize=14)
    fig.tight_layout()
    plt.show()
```
